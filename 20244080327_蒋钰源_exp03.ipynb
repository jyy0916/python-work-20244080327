```python
import re

text3 = "ct cat caat caaat"

print("ca*t    ->", re.findall(r"ca*t", text3))
print("ca+t    ->", re.findall(r"ca+t", text3))
print("ca?t    ->", re.findall(r"ca?t", text3))
print("ca{2}t  ->", re.findall(r"ca{2}t", text3))
print("ca{1,2}t->", re.findall(r"ca{1,2}t", text3))
```

在文本 `"ct cat caat caaat"` 上分别使用不同量词，匹配结果与含义如下：

- **`ca*t`**：`*` 表示前面的字符 `a` 出现 **0 次或多次**
- **`ca+t`**：`+` 表示前面的字符 `a` 出现 **1 次或多次**
- **`ca?t`**：`?` 表示前面的字符 `a` 出现 **0 次或 1 次**
- **`ca{2}t`**：`{2}` 表示前面的字符 `a` **恰好出现 2 次**
- **`ca{1,2}t`**：`{1,2}` 表示前面的字符 `a` 出现 **1 到 2 次**

```python
import re

print("=" * 30, "任务1：提取邮箱用户名和域名", "=" * 30)
emails = "联系: alice@example.com; 备选: bob@test.org"
email_pairs = re.findall(r"([\w.-]+)@([\w.-]+\.\w+)", emails)
for i, (user, domain) in enumerate(email_pairs, 1):
    print(f"邮箱 {i} -> 用户名: {user:5s} | 域名: {domain}")

print("\n" + "=" * 30, "任务2：找出连续重复的单词", "=" * 30)
dup_text = "cat cat, dog cat, bird bird"
duplicates = re.findall(r"\b(\w+)\s+\1\b", dup_text)
print("发现重复的单词有:", duplicates)

for m in re.finditer(r"\b(\w+)\s+\1\b", dup_text):
    print(f"完整匹配片段: '{m.group(0)}' (起始位置: {m.start()})")

print("\n" + "=" * 30, "任务3：日期格式转换", "=" * 30)
date_text = "2024-05-01\n2024/05/02\n2024.05.03"
print("【转换前】:\n" + date_text)

date_new = re.sub(r"(\d{4})[-/.](\d{2})[-/.](\d{2})", r"\2-\3-\1", date_text)
print("\n【转换后】:\n" + date_new)

print("\n" + "=" * 30, "任务4：电话号码统一格式化", "=" * 30)
phone_text = "联系我: 123-456-7890 或 987.654.3210"
print("【格式化前】:", phone_text)

phone_new = re.sub(r"(\d{3})[-.](\d{3})[-.](\d{4})", r"(\1) \2-\3", phone_text)
print("【格式化后】:", phone_new)
```

1. **提取邮箱**：使用捕获组 `([\w.-]+)@([\w.-]+\.\w+)`，第一组提取用户名，第二组提取域名。
2. **连续重复单词**：使用 `\b(\w+)\s+\1\b`，其中 `\1` 是回溯引用，表示与第一个捕获组完全相同的单词。
3. **日期转换**：使用 `(\d{4})[-/.](\d{2})[-/.](\d{2})` 匹配年、月、日，并在替换表达式中使用 `\2-\3-\1` 将顺序调换为“月-日-年”。
4. **电话格式化**：使用 `(\d{3})[-.](\d{3})[-.](\d{4})` 匹配不同分隔符的电话号码，在替换表达式中使用 `(\1) \2-\3` 将其统一格式。

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲: 正则表达式</p>
<p class="course">第七讲: 结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

print("=" * 30, "任务1：获取标题和 h1", "=" * 30)
print("页面标题文本:", soup.title.get_text(strip=True))
print("h1 标签名:", soup.h1.name)
print("h1 文本:", soup.h1.get_text(strip=True))


print("\n" + "=" * 30, "任务2：查找课程段落和所有链接", "=" * 30)
course_ps = soup.find_all("p", class_="course")
print(f"找到 {len(course_ps)} 个课程段落:")
for p in course_ps:
    print(" -", p.get_text(strip=True))

links = soup.find_all("a")
print(f"\n找到 {len(links)} 个链接:")
for a in links:
    print(f" - 链接文字: {a.get_text(strip=True):4s} | href属性: {a.get('href')}")


print("\n" + "=" * 30, "任务3：比较 get_text() 的结果", "=" * 30)
print("【整个文档的文本】:")
print(repr(soup.get_text()))

print("\n【单个段落的文本】:")
print(repr(course_ps[0].get_text()))

print("\n【单个链接的文本】:")
print(repr(links[0].get_text()))
```

1. 使用 `BeautifulSoup(html_doc, "html.parser")` 构建解析树，可以直接通过 `soup.title` 和 `soup.h1` 访问对应的标签，使用 `.name` 获取标签名，使用 `.get_text(strip=True)` 获取干净的文本。
2. `find_all("p", class_="course")` 用于查找所有指定 class 的段落，`find_all("a")` 用于查找所有超链接。对于标签对象，使用 `.get("href")` 可以提取属性值。
3. `get_text()` 用于提取标签内的所有可见文本。整个文档的 `get_text()` 会包含标签之间的换行和大量空白，而单个段落或链接的 `get_text()` 只会提取该标签内部的文本，范围更小、更精确。
