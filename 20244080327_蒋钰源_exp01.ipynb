```python
print(" 任务 1：score_level 函数 ")
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在 0 到 100 之间！")
    elif score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for s in [95, 85, 70, 50]:
    print(f"分数 {s}: {score_level(s)}")

try:
    score_level(105)
except ValueError as e:
    print(f"捕获异常：{e}")

print("\n 任务 2：1 到 100 求和 ")
sum_for = 0
for i in range(1, 101):
    sum_for += i

sum_while = 0
j = 1
while j <= 100:
    sum_while += j
    j += 1

print(f"for 循环结果: {sum_for}")
print(f"while 循环结果: {sum_while}")
print(f"结果是否一致: {sum_for == sum_while}")

print("\n 任务 3：阶乘的递归与迭代 ")
def factorial_recursive(n):
    if n == 0:
        return 1
    else:
        return n * factorial_recursive(n - 1)

def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

for n in [0, 1, 5, 10]:
    rec = factorial_recursive(n)
    ite = factorial_iterative(n)
    print(f"n={n}: 递归={rec}, 迭代={ite}, 一致={rec == ite}")

print("\n 任务 4：summary 函数 ")
def summary(values):
    return (min(values), max(values), sum(values) / len(values))

data = [85, 92, 78, 60, 95]

min_val, max_val, avg_val = summary(data)
print(f"数据: {data}")
print(f"最小值: {min_val}, 最大值: {max_val}, 平均值: {avg_val:.2f}")

print("\n 任务 5：sum_all 可变参数 ")
def sum_all(*args):
    return sum(args)

print(f"无参数: {sum_all()}")
print(f"单参数: {sum_all(10)}")
print(f"多参数: {sum_all(1, 2, 3, 4, 5)}")
```

1.分数等级主要是用 if-elif-else 分情况。如果分数不在 0~100 之间就报 ValueError 错误。

2.阶乘递归的时候注意 n=0 时要直接返回 1，不然停不下来。迭代就是用 for 循环乘，两种方法结果一样。

3.sum_all 用了 *args 接收任意个参数，测试了传 0 个、1 个和多个的情况，直接 return sum(args) 。

```python
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

print(" 任务1：zip构建字典 ")

score_dict = dict(zip(names, scores))
for name, score in score_dict.items():
    print(f"{name}: {score}")

print("\n 任务2：enumerate 输出序号 ")
for i, (name, score) in enumerate(score_dict.items(), start=1):
    print(f"序号{i}: {name}, 成绩{score}")

print("\n 任务3：列表推导式 ")
passed_names = [name for name, score in score_dict.items() if score >= 60]
print(f"及格学生: {passed_names}")
squared_scores = [score ** 2 for score in scores]
print(f"成绩平方: {squared_scores}")

print("\n 任务4：基础统计 ")
count = len(scores)
avg_score = sum(scores) / count
max_score = max(scores)
min_score = min(scores)
pass_rate = len(passed_names) / count

print(f"人数: {count}")
print(f"平均分: {avg_score:.2f}")
print(f"最高分: {max_score}, 最低分: {min_score}")
print(f"及格率: {pass_rate:.1%}")

print("\n 任务5：引用与浅拷贝（重点） ")
demo_scores = [10, 20, 30]
alias = demo_scores
copied = demo_scores.copy()

alias[0] = 999

print(f"原列表 demo_scores: {demo_scores}")
print(f"alias 修改后: {alias}")
print(f"copy 列表: {copied}")

print("\n 任务6：元组不可变与 zip 截断 ")
student_record = ("Alice", 85)
try:
    student_record[1] = 90
except TypeError as e:
    print(f"元组不能直接赋值: {e}")

list_a = [1, 2, 3, 4]
list_b = ['a', 'b']
print(f"不等长 zip 结果: {list(zip(list_a, list_b))}")
```

直接赋值（alias = scores）只是给原列表起了个别名，修改 alias 会导致原列表跟着变。而 copy() 是浅拷贝，修改 copied 不会影响原列表。
元组一旦创建就不能改了，硬改会报 TypeError。另外，zip 如果遇到长度不一样的序列，会以短的为准，后面长出来的部分直接忽略。

```python
import copy

print(" 任务1 & 2：定义 Student 类并创建对象 ")
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):

        status = "及格" if self.is_passed() else "不及格"
        return f"学生: {self.name}, 成绩: {self.score} ({status})"

student_objs = []
for record in valid_records:
    obj = Student(record["name"], record["score"])
    student_objs.append(obj)

for obj in student_objs:
    print(obj)


print("\n 任务3：定义 GraduateStudent 子类 ")
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 调用父类初始化
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"研究生: {self.name}, 成绩: {self.score} ({status}), 研究方向: {self.research_topic}"

grad_stu = GraduateStudent("张三", 92, "人工智能")
print(grad_stu)


print("\n 任务4：直接赋值、浅拷贝与深拷贝 ")

original = {"name": "Alice", "scores": [85, 90]}

print(f"原始对象: {original}")

alias_obj = original
alias_obj["scores"].append(100)
print(f"直接赋值修改后: {original}")

original = {"name": "Alice", "scores": [85, 90]}

shallow_copy = copy.copy(original)
shallow_copy["scores"].append(99)
print(f"浅拷贝修改后: {original}")

original = {"name": "Alice", "scores": [85, 90]}

deep_copy = copy.deepcopy(original)
deep_copy["scores"].append(77)
print(f"深拷贝修改后: {original}")
```


1. 定义了 `Student` 类，初始化了名字和分数。`is_passed()` 判断是否大于等于 60，`__str__()` 用来规范打印出来的格式。
2. 定义 `GraduateStudent` 继承自 `Student`，用 `super().__init__()` 初始化父类的属性，同时重写了 `__str__()`，把研究方向也打印出来了。
3. 重点关于复制：
   直接赋值（`alias = original`）只是换了个名字，改哪个另一个都会跟着变。
   `copy.copy()` 是浅拷贝，它只复制了最外层（字典），里面的列表还是共享的。所以修改列表后，原数据也变了。
   `copy.deepcopy()` 是深拷贝，它连里面的列表都全新复制了一份。修改深拷贝的数据，对原数据毫无影响。
