```python
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m

print("GCD(2018, 2019) =", gcd(2018, 2019))
print("GCD(1200, 300) =", gcd(1200, 300))
print("GCD(5040, 60) =", gcd(5040, 60))

print("\n 负数测试 ")
print("GCD(-10, 5) =", gcd(-10, 5))
print("GCD(10, -5) =", gcd(10, -5))
```

如果参数包含负数，Python 中的 % 运算结果符号会直接由除数决定。询问大数据ai模型后告知没有数学意义，如果要使其有意义，应该在函数开头添加显式的负数检查并抛出异常。

```python
import math

def euler_limit(n):
    return (1 + 1/n) ** n

def euler_infinite_sum(n):
    if n == 0:
        return 0.0

    total = 0.0
    fact = 1.0
    for k in range(n):
        if k > 0:
            fact *= k
        total += 1.0 / fact
    return total

def euler_approx(epsilon):
    true_e = math.exp(1)
    total = 0.0
    fact = 1.0
    k = 0
    while abs(total - true_e) > epsilon:
        if k > 0:
            fact *= k
        total += 1.0 / fact
        k += 1
    return total

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"n={k}: {euler_infinite_sum(k)}")

print("1. 测试 euler_limit")
print("euler_limit(10) =", euler_limit(10))
print("euler_limit(1000) =", euler_limit(1000))

print("\n 2. 测试 euler_infinite_sum")
print("euler_infinite_sum(0) =", euler_infinite_sum(0))
print("euler_infinite_sum(1) =", euler_infinite_sum(1))
print("euler_infinite_sum(4) =", euler_infinite_sum(4))

print("\n 3. 测试 euler_approx")
print("euler_approx(0.1) =", euler_approx(0.1))
print("euler_approx(0.0001) =", euler_approx(0.0001))

print("\n 4. 测试 print_euler_sum_table")
print_euler_sum_table(5)
```

```python
def is_power_of_2(n):
    if n <= 0:
        return False

    while n % 2 == 0:
        n //= 2
    return n == 1

def is_power(b, n):
    if n <= 0:
        return False
    if b == 1:
        return n == 1

    while n % b == 0:
        n //= b

    return n == 1

print(" 测试 is_power_of_2 ")
print("is_power_of_2(8) =", is_power_of_2(8))
print("is_power_of_2(10) =", is_power_of_2(10))
print("is_power_of_2(1) =", is_power_of_2(1))
print("is_power_of_2(0) =", is_power_of_2(0))

print("\n 测试 is_power ")
print("is_power(3, 27) =", is_power(3, 27))
print("is_power(2, 10) =", is_power(2, 10))
print("is_power(5, 125) =", is_power(5, 125))
print("is_power(4, 16) =", is_power(4, 16))
```
