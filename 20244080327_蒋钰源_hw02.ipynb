```python
print(" 1. 字符串")

def is_palindrome(s):

    cleaned = [char.lower() for char in s if char.isalnum()]

    return cleaned == cleaned[::-1]

def is_abecedarian(s):
    cleaned = [char.lower() for char in s if char.isalpha()]

    for i in range(1, len(cleaned)):
        if cleaned[i] < cleaned[i-1]:
            return False
    return True

def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return "".join([char for char in s if char not in vowels])

print("is_palindrome('T A C O cat'):", is_palindrome("T A C O cat"))
print("is_palindrome('hello'):", is_palindrome("hello"))
print("is_abecedarian('abcd efgh xyz'):", is_abecedarian("abcd efgh xyz"))
print("is_abecedarian('dog'):", is_abecedarian("dog"))
print("remove_vowels('audio'):", remove_vowels("audio"))


print("\n 2. 列表")

def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表！")
    return lst[::-1]

def is_sorted(seq):

    for a, b in zip(seq, seq[1:]):
        if a > b:
            return False
    return True

print("list_reverse([1, 2, 3]):", list_reverse([1, 2, 3]))
try:
    list_reverse("not a list")
except TypeError as e:
    print("类型检查捕获异常:", e)
print("is_sorted([1, 2, 2, 3]):", is_sorted([1, 2, 2, 3]))
print("is_sorted([1, 3, 2]):", is_sorted([1, 3, 2]))
```

1. 回文判断先把字母数字留下来了，然后统一转小写，最后用切片 `[::-1]` 翻转对比，直接返回 True/False。
2. 字母顺序判断也是先去掉空格，然后在循环里只要发现后一个字母比前一个字母小，立马 return False。
3. 去掉元音就是用个字符串装 aeiouAEIOU这些元音字符，把不在里面的字符重新拼起来。
4. 反转列表用切片 `lst[::-1]` 最方便，不改变原列表。又加个 `isinstance` 检查，防止乱传字符串进来。
5. 判断升序这题要求只能遍历一次，用了 `zip(seq, seq[1:])` 把相邻的两个元素拼在一起比较。

```python
print(" 3. 作为向量的元组 ")

def vec_scalar_mult(t, s):
    if not isinstance(t, tuple) or not isinstance(s, (int, float)):
        raise TypeError("参数必须是一个元组和一个数字！")
    return tuple(x * s for x in t)

def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组！")
    if len(t1) != len(t2):
        raise ValueError("两个向量长度必须一致！")
    return sum(x * y for x, y in zip(t1, t2))

def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return False

    for row in mx:
        if not isinstance(row, tuple):
            return False

        for val in row:
            if not isinstance(val, (int, float)):
                return False

    first_len = len(mx[0])
    return all(len(row) == first_len for row in mx)

def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise ValueError("矩阵格式不正确！")
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组！")
    if len(mx[0]) != len(v):
        raise ValueError("矩阵列数必须等于向量长度！")

    result = []
    for row in mx:
        result.append(sum(x * y for x, y in zip(row, v)))
    return tuple(result)

print("vec_scalar_mult((1, 2, 3), 2):", vec_scalar_mult((1, 2, 3), 2))
print("vec_inner_product((1, 2), (3, 4)):", vec_inner_product((1, 2), (3, 4)))
mx = ((1, 2), (3, 4), (5, 6))
print("check_valid_mx(mx):", check_valid_mx(mx))
print("mx_vec_mult(mx, (1, 1)):", mx_vec_mult(mx, (1, 1)))
```

元组与矩阵思路记录：
1. 数乘用生成器表达式乘完再转回元组。`isinstance` 加条件把非法输入拦截。
2. 内积用 `zip` 打包一起算，最后 `sum`。加上了长度检查和元组类型检查。
3. `check_valid_mx` 负责看矩阵是不是元组的元组，里面的元素是不是数字，还有每行长度是不是一样的。
4. `mx_vec_mult` 先调校验函数，然后每一行和向量做内积，打包成元组返回。

```python
from functools import total_ordering

@total_ordering
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or dim < 0:
            raise ValueError("维度必须是非负整数！")
        self.dim = dim

        if elements is None:

            self.elements = tuple(0.0 for _ in range(dim))
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("元素必须是列表或元组！")
            if len(elements) != dim:
                raise ValueError("元素个数必须和维度一致！")
            self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    def __eq__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和 Vector 比较！")
        if self.dim != other.dim:
            raise ValueError("不同维度的向量不能比较！")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和 Vector 比较！")
        if self.dim != other.dim:
            raise ValueError("不同维度的向量不能比较！")
        return self.elements < other.elements

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象！")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法计算内积！")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):

        if isinstance(other, (int, float)):
            new_elements = tuple(x * other for x in self.elements)
            return Vector(self.dim, new_elements)

        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度必须一致！")
            new_elements = tuple(a * b for a, b in zip(self.elements, other.elements))
            return Vector(self.dim, new_elements)
        else:
            raise TypeError("只能乘标量或向量！")


    def __rmul__(self, other):
        return self.__mul__(other)

print("--- Vector 测试 ---")
v1 = Vector(2, (2, 3))
v2 = Vector(2, (4, 5))
v3 = Vector(2, (2, 5))

print("v1:", v1)
print("v1 + 标量乘 2:", 2 * v1)
print("v1 向量乘 v2:", v1 * v2)
print("v1 < v3:", v1 < v3)
print("v1.dot(v2):", v1.dot(v2))

try:
    v1 < Vector(3, (1, 2, 3))
except ValueError as e:
    print("比较不同维度报错:", e)
```

1. 初始化时检查了维度（必须是非负整数），元素不给的话默认全填 0。
2. 维度和元素选了**实例属性**。因为每个 Vector 的维度大小和元素都不一样，如果放成类属性，所有对象会共享同一份，改了一个全变。
3. 比较运算符用了 `functools.total_ordering` 装饰器，只要自己写 `__eq__` (相等) 和 `__lt__` (小于)，大于、小于等于都会自动生成。比较前查了维度，不一样就报错。
4. `dot` 方法先检查是不是 Vector，再查维度是不是一样，最后算内积。
5. 乘法 `__mul__` 同时兼容了数字和向量两种情况，如果是数字左乘（`2 * v`），用 `__rmul__` 指回去。
